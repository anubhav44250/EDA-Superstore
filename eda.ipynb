```python
# Exploratory Data Analysis on Superstore Sales Dataset

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

""" Objective
The goal of this project is to explore the Superstore dataset to understand sales, profit, and customer patterns using Python visualization and analysis tools."""


sns.set_theme(style="darkgrid")

df = pd.read_csv("C:\CSE\Projects\Exploratory Data Analysis on Superstore Sales Dataset\data\superstore.csv")
df.head()
df.shape
df.columns
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
df = df.drop_duplicates()


sns.histplot(df["Sales"], kde=True)
plt.title("Sales Distribution")
plt.show()

#Sales distribution is right-skewed. Most orders are small with few high-value sales.


sns.histplot(df["Profit"], kde=True)
plt.title("Profit Distribution")
plt.show 

#Profit contains both positive and negative values, indicating some loss-making orders.


sns.boxplot(x=df["Sales"])
plt.show()

sns.countplot(x="Category", data = df)
plt.show()

plt.figure(figsize=(14,6))
sns.countplot(x="Region", data=df)
plt.xticks(rotation=45, ha="right")
plt.show 

plt.figure(figsize=(6,5))
sns.scatterplot(x="Sales", y="Profit", hue="Category", data=df, alpha=0.6)
plt.title("Sales vs Profit by Category")
plt.show()

sns.boxplot(x="Category", y = "Sales", data = df)
plt.show()

plt.figure(figsize=(14,6))
sns.boxplot(x="Region", y ="Profit", data=df)
plt.xticks(rotation=45 , ha="right")
plt.title("Profit Distribution by Region")
plt.tight_layout()
plt.show()

corr = df.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True ,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
